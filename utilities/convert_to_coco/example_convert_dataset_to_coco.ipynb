```python
import os
import shutil

def move_vids_to_backup(base_dataset_path, vids_to_move, backup_dir_name="backup"):
    """
    Move specified VID folders to a backup directory
    
    Args:
        base_dataset_path: Path to cholecinstanceseg_reference_image_and_annotations
        vids_to_move: List of VID numbers to move
        backup_dir_name: Name of backup directory to create
    """
    
    # Create backup directory at the same level as the dataset
    backup_path = os.path.join(os.path.dirname(base_dataset_path), backup_dir_name)
    os.makedirs(backup_path, exist_ok=True)
    
    # Create train, val, test subdirectories in backup
    for split in ['train', 'val', 'test']:
        os.makedirs(os.path.join(backup_path, split), exist_ok=True)
    
    moved_folders = []
    not_found_vids = []
    
    # Process each split
    for split in ['train', 'val', 'test']:
        split_path = os.path.join(base_dataset_path, split)
        
        if not os.path.exists(split_path):
            print(f"Split {split} not found, skipping...")
            continue
            
        print(f"\n=== Processing {split.upper()} split ===")
        
        # Get all folders in this split
        all_folders = [f for f in os.listdir(split_path) 
                      if os.path.isdir(os.path.join(split_path, f))]
        
        for folder in all_folders:
            # Extract VID number from folder name
            # Folders are typically named like: VID01_seg8k, VID02_t50_sparse, etc.
            if folder.startswith('VID'):
                try:
                    # Extract the number part (e.g., "01" from "VID01_seg8k")
                    vid_str = folder[3:].split('_')[0]
                    vid_num = int(vid_str)
                    
                    if vid_num in vids_to_move:
                        source_path = os.path.join(split_path, folder)
                        dest_path = os.path.join(backup_path, split, folder)
                        
                        try:
                            shutil.move(source_path, dest_path)
                            moved_folders.append(f"{split}/{folder}")
                            print(f"  ✓ Moved: {folder}")
                        except Exception as e:
                            print(f"  ✗ Error moving {folder}: {str(e)}")
                            
                except (ValueError, IndexError):
                    print(f"  ? Could not parse VID number from: {folder}")
    
    # Check which VIDs were not found
    found_vids = set()
    for folder_path in moved_folders:
        folder_name = os.path.basename(folder_path)
        if folder_name.startswith('VID'):
            try:
                vid_str = folder_name[3:].split('_')[0]
                vid_num = int(vid_str)
                found_vids.add(vid_num)
            except (ValueError, IndexError):
                pass
    
    not_found_vids = set(vids_to_move) - found_vids
    
    # Print summary
    print(f"\n=== SUMMARY ===")
    print(f"Total folders moved: {len(moved_folders)}")
    print(f"Backup directory: {backup_path}")
    
    if moved_folders:
        print(f"\nMoved folders:")
        for folder in sorted(moved_folders):
            print(f"  - {folder}")
    
    if not_found_vids:
        print(f"\nVIDs not found in dataset:")
        for vid in sorted(not_found_vids):
            print(f"  - VID{vid:02d}")

# List of VID numbers to move
vids_to_move = [2, 4, 5, 6, 8, 10, 13, 31, 32, 36, 40, 50, 42, 47, 49, 51, 56, 57, 60, 62, 41, 42, 43, 44, 45]

# Remove duplicates and sort
vids_to_move = sorted(list(set(vids_to_move)))
print(f"VIDs to move: {vids_to_move}")

# Set your dataset path
base_dataset_path = "/home/yang/Downloads/instance_seg_dataset/cholecinstanceseg_reference_image_and_annotations"

# Move the folders
move_vids_to_backup(base_dataset_path, vids_to_move)
```

```python
import os
import sys
from pathlib import Path

# Add the utilities path to sys.path
sys.path.append('cholec_instance_seg')

from utilities.convert_to_coco.convert_dataset_to_coco_utils import annotations_to_coco

def convert_all_datasets_to_coco(base_dataset_path, output_base_dir):
    """
    Convert all dataset folders in train/val splits to COCO format
    
    Args:
        base_dataset_path: Path to cholecinstanceseg_reference_image_and_annotations
        output_base_dir: Directory to save all COCO JSON files
    """
    
    # Create output directory
    os.makedirs(output_base_dir, exist_ok=True)
    
    # Iterate through train and val splits
    for split in ['train', 'val']:
        split_path = os.path.join(base_dataset_path, split)
        
        if not os.path.exists(split_path):
            print(f"Split {split} not found at {split_path}")
            continue
            
        print(f"\n=== Processing {split.upper()} split ===")
        
        # Create output directory for this split
        split_output_dir = os.path.join(output_base_dir, split)
        os.makedirs(split_output_dir, exist_ok=True)
        
        # Get all video folders in this split
        video_folders = sorted([f for f in os.listdir(split_path) 
                               if os.path.isdir(os.path.join(split_path, f))])
        
        print(f"Found {len(video_folders)} video folders in {split}")
        
        # Process each video folder
        for i, video_folder in enumerate(video_folders):
            video_path = os.path.join(split_path, video_folder)
            
            # Check if this folder has img_dir and ann_dir
            img_dir = os.path.join(video_path, 'img_dir')
            ann_dir = os.path.join(video_path, 'ann_dir')
            
            if not (os.path.exists(img_dir) and os.path.exists(ann_dir)):
                print(f"  Skipping {video_folder}: missing img_dir or ann_dir")
                continue
            
            # Define output path for this video's COCO annotations
            output_json_path = os.path.join(split_output_dir, f"{video_folder}_coco.json")
            
            try:
                print(f"  [{i+1}/{len(video_folders)}] Converting {video_folder}...")
                
                # Convert to COCO format
                annotations_to_coco(video_path, output_json_path, vis=False)
                
                print(f"    ✓ Saved: {output_json_path}")
                
            except Exception as e:
                print(f"    ✗ Error converting {video_folder}: {str(e)}")
                continue

def convert_single_split_to_combined_coco(base_dataset_path, split, output_json_path):
    """
    Convert all videos in a single split to one combined COCO file
    
    Args:
        base_dataset_path: Path to cholecinstanceseg_reference_image_and_annotations
        split: 'train' or 'val'
        output_json_path: Path to save combined COCO JSON
    """
    import json
    from utilities._general_utils.dataset_variables import CholecInstanceSegVariables
    from utilities.convert_to_coco.convert_dataset_to_coco_utils import initalize_coco_annotation, mask_or_contour_to_coco
    
    split_path = os.path.join(base_dataset_path, split)
    
    if not os.path.exists(split_path):
        print(f"Split {split} not found at {split_path}")
        return
    
    # Initialize combined COCO annotation
    instrument_id_to_instrument_class_dict = CholecInstanceSegVariables.instrument_id_to_instrument_class_dict
    combined_coco = initalize_coco_annotation(instrument_id_to_instrument_class_dict)
    
    image_id_counter = 1
    video_folders = sorted([f for f in os.listdir(split_path) 
                           if os.path.isdir(os.path.join(split_path, f))])
    
    print(f"Combining {len(video_folders)} videos from {split} split...")
    
    for video_folder in video_folders:
        video_path = os.path.join(split_path, video_folder)
        img_dir = os.path.join(video_path, 'img_dir')
        ann_dir = os.path.join(video_path, 'ann_dir')
        
        if not (os.path.exists(img_dir) and os.path.exists(ann_dir)):
            continue
            
        print(f"  Processing {video_folder}...")
        
        # Get all images in this video
        img_files = sorted([f for f in os.listdir(img_dir) if f.endswith('.png')])
        ann_files = sorted([f for f in os.listdir(ann_dir) if f.endswith('.json')])
        
        for img_file, ann_file in zip(img_files, ann_files):
            # Add image info
            image_info = {
                "file_name": f"{split}/{video_folder}/img_dir/{img_file}",  # Include video folder in filename
                "id": image_id_counter,
                "width": 854,
                "height": 480,
            }
            combined_coco["images"].append(image_info)
            
            # Convert annotations for this image
            try:
                mask_or_contour_to_coco(ann_file, ann_dir, image_id_counter, combined_coco, vis=False)
            except Exception as e:
                print(f"    Error processing {ann_file}: {e}")
            
            image_id_counter += 1
    
    # Save combined COCO file
    os.makedirs(os.path.dirname(output_json_path), exist_ok=True)
    with open(output_json_path, 'w') as f:
        json.dump(combined_coco, f)
    
    print(f"Combined COCO file saved: {output_json_path}")
    print(f"Total images: {len(combined_coco['images'])}")
    print(f"Total annotations: {len(combined_coco['annotations'])}")

# Usage examples:

# # Option 1: Convert each video folder to separate COCO files
# base_dataset_path = "/home/yang/Downloads/instance_seg_dataset/cholecinstanceseg_reference_image_and_annotations"
# output_dir = "/home/yang/Downloads/instance_seg_dataset/coco_annotations_separate"

# convert_all_datasets_to_coco(base_dataset_path, output_dir)

# Option 2: Convert each split to a single combined COCO file
output_dir_combined = "/home/yang/Downloads/instance_seg_dataset/coco_annotations_combined"
os.makedirs(output_dir_combined, exist_ok=True)

convert_single_split_to_combined_coco(
    base_dataset_path, 
    'train', 
    os.path.join(output_dir_combined, 'train_combined_coco.json')
)

convert_single_split_to_combined_coco(
    base_dataset_path, 
    'val', 
    os.path.join(output_dir_combined, 'val_combined_coco.json')
)
```

```python
import json
import os
import shutil
from pathlib import Path
import cv2

def convert_coco_to_yolo(coco_json_path, images_base_dir, output_dir, split_name):
    """
    Convert COCO format annotations to YOLO format
    
    Args:
        coco_json_path: Path to COCO JSON file
        images_base_dir: Base directory containing the images
        output_dir: Output directory for YOLO dataset
        split_name: 'train' or 'val'
    """
    
    # Load COCO annotations
    with open(coco_json_path, 'r') as f:
        coco_data = json.load(f)
    
    # Create output directories
    yolo_dir = Path(output_dir)
    images_dir = yolo_dir / "images" / split_name
    labels_dir = yolo_dir / "labels" / split_name
    
    images_dir.mkdir(parents=True, exist_ok=True)
    labels_dir.mkdir(parents=True, exist_ok=True)
    
    # Create class mapping
    categories = {cat['id']: cat['name'] for cat in coco_data['categories']}
    class_names = [categories[i] for i in sorted(categories.keys())]
    
    # Create images and annotations mapping
    images = {img['id']: img for img in coco_data['images']}
    annotations_by_image = {}
    
    for ann in coco_data['annotations']:
        image_id = ann['image_id']
        if image_id not in annotations_by_image:
            annotations_by_image[image_id] = []
        annotations_by_image[image_id].append(ann)
    
    # Process each image
    for img_id, img_info in images.items():
        # Copy image file
        original_img_path = Path(images_base_dir) / img_info['file_name']
        new_img_name = f"{Path(img_info['file_name']).stem}.jpg"
        new_img_path = images_dir / new_img_name
        
        if original_img_path.exists():
            # Convert to JPG if needed
            img = cv2.imread(str(original_img_path))
            cv2.imwrite(str(new_img_path), img)
        else:
            print(f"Warning: Image not found: {original_img_path}")
            continue
        
        # Create YOLO label file
        label_path = labels_dir / f"{Path(img_info['file_name']).stem}.txt"
        
        img_width = img_info['width']
        img_height = img_info['height']
        
        yolo_annotations = []
        
        if img_id in annotations_by_image:
            for ann in annotations_by_image[img_id]:
                # Get bounding box
                bbox = ann['bbox']  # [x, y, width, height]
                x, y, w, h = bbox
                
                # Convert to YOLO format (normalized center coordinates)
                x_center = (x + w / 2) / img_width
                y_center = (y + h / 2) / img_height
                norm_width = w / img_width
                norm_height = h / img_height
                
                # Get class ID (YOLO uses 0-based indexing)
                class_id = ann['category_id'] - 1  # Convert from 1-based to 0-based
                
                yolo_annotations.append(f"{class_id} {x_center:.6f} {y_center:.6f} {norm_width:.6f} {norm_height:.6f}")
        
        # Write label file
        with open(label_path, 'w') as f:
            f.write('\n'.join(yolo_annotations))
    
    return class_names

def create_yolo_dataset(base_dataset_path, coco_annotations_dir, output_dir):
    """
    Create complete YOLO dataset from COCO annotations
    """
    
    # Convert train and val splits
    train_coco_path = os.path.join(coco_annotations_dir, "train_combined_coco.json")
    val_coco_path = os.path.join(coco_annotations_dir, "val_combined_coco.json")
    
    print("Converting training data...")
    class_names = convert_coco_to_yolo(
        train_coco_path, 
        base_dataset_path, 
        output_dir, 
        "train"
    )
    
    print("Converting validation data...")
    convert_coco_to_yolo(
        val_coco_path, 
        base_dataset_path, 
        output_dir, 
        "val"
    )
    
    # Create data.yaml file
    yaml_content = f"""# YOLO Dataset Configuration
path: {os.path.abspath(output_dir)}
train: images/train
val: images/val

# Number of classes
nc: {len(class_names)}

# Class names
names: {class_names}
"""
    
    yaml_path = os.path.join(output_dir, "data.yaml")
    with open(yaml_path, 'w') as f:
        f.write(yaml_content)
    
    print(f"YOLO dataset created at: {output_dir}")
    print(f"Classes: {class_names}")
    print(f"Configuration saved to: {yaml_path}")
    
    return yaml_path

# Usage
base_dataset_path = "/home/yang/Downloads/instance_seg_dataset/cholecinstanceseg_reference_image_and_annotations"
coco_annotations_dir = "/home/yang/Downloads/instance_seg_dataset/coco_annotations_combined"
yolo_output_dir = "/home/yang/Downloads/instance_seg_dataset/yolo_dataset"

yaml_path = create_yolo_dataset(base_dataset_path, coco_annotations_dir, yolo_output_dir)
```
